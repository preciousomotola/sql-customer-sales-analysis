###Customer Analysis

```python
pd.read_sql_query("""
SELECT
    customers.customer_name,
        COUNT(orders.order_id) AS total_orders,
            SUM(orders.revenue) AS total_revenue,
                ROUND(AVG(orders.revenue), 2) AS average_order_value
                FROM orders
                INNER JOIN customers
                    ON orders.customer_id = customers.customer_id
                    GROUP BY customers.customer_name
                    ORDER BY total_revenue DESC;
                    """, conn)
```

### Customer Performance Insight

Ada generated the highest total revenue at ₦295,000 and also had the highest number of orders with 4 purchases.

Emeka generated ₦285,000 from 3 orders and had the highest average order value at ₦95,000, indicating that Emeka makes fewer but higher-value purchases.

Bola had the lowest average order value at approximately ₦48,333, suggesting an opportunity to increase purchase value through targeted offers or product recommendations.

### High-Value Customers

Customers were classified as high-value when they had at least 3 orders and generated more than ₦180,000 in total revenue.

```python
pd.read_sql_query("""
SELECT
    customers.customer_name,
        COUNT(orders.order_id) AS total_orders,
            SUM(orders.revenue) AS total_revenue
            FROM orders
            INNER JOIN customers
                ON orders.customer_id = customers.customer_id
                GROUP BY customers.customer_name
                HAVING
                    COUNT(orders.order_id) >= 3
                        AND SUM(orders.revenue) > 180000
                        ORDER BY total_revenue DESC;
                        """, conn)
```

###Product Analysis

This section evaluates product performance based on order volume, total revenue, and average order value to identify the strongest products and potential growth opportunities.

```python
pd.read_sql_query("""
SELECT
    product,
        COUNT(order_id) AS total_orders,
            SUM(revenue) AS total_revenue,
                ROUND(AVG(revenue), 2) AS average_order_value
                FROM orders
                GROUP BY product
                ORDER BY total_revenue DESC;
                """, conn)
```

### Product Performance Insight

Rice was the strongest-performing product, generating ₦485,000 in revenue from 6 orders and recording the highest average order value of approximately ₦80,833.

Beans generated ₦295,000 from 5 orders, while Oil generated ₦260,000 from 4 orders.

Although Oil generated less total revenue than Beans, its average order value was higher at ₦65,000 compared with Beans at ₦59,000. This suggests that Oil may have potential for growth if its order volume can be increased.

Management should maintain Rice's strong performance while exploring strategies to improve the sales volume of Beans and Oil.

```python
pd.read_sql_query("""
SELECT
    product,
        SUM(revenue) AS total_revenue,
            ROUND(
                    SUM(revenue) * 100.0 /
                            (SELECT SUM(revenue) FROM orders),
                                    2
                                        ) AS revenue_contribution_percent
                                        FROM orders
                                        GROUP BY product
                                        ORDER BY total_revenue DESC;
                                        """, conn)
```

### Product Revenue Contribution Insight

Rice contributed the largest share of total revenue at 46.63%, making it the leading product in the dataset.

Beans contributed 28.37%, while Oil contributed 25.00%.

Although Rice generated almost half of total revenue, the other two products together contributed 53.37%. This shows that the business is not completely dependent on Rice and has opportunities to strengthen Beans and Oil.

Management should maintain Rice's strong performance while developing strategies to increase the revenue contribution of Beans and Oil.

###Monthly Sales Analysis

This section examines monthly revenue performance to identify sales trends and changes in average order value over time.

```python
pd.read_sql_query("""
SELECT
    SUBSTR(order_date, 1, 7) AS month,
        COUNT(order_id) AS total_orders,
            SUM(revenue) AS total_revenue,
                ROUND(AVG(revenue), 2) AS average_order_value
                FROM orders
                GROUP BY SUBSTR(order_date, 1, 7)
                ORDER BY month;
                """, conn)
```

### Monthly Sales Insight

Monthly revenue increased consistently from ₦290,000 in January to ₦355,000 in February and ₦395,000 in March.

The number of orders remained constant at 5 orders each month, while average order value increased from ₦58,000 in January to ₦71,000 in February and ₦79,000 in March.

This indicates that revenue growth was driven by higher-value purchases rather than an increase in order volume.

Management should investigate the factors behind the increase in average order value and consider strategies that encourage customers to make larger purchases.

###Regional Analysis

This section evaluates revenue performance across regions to identify regional strengths, customer concentration, and potential growth opportunities.

```python
pd.read_sql_query("""
SELECT
    customers.region,
        COUNT(DISTINCT customers.customer_id) AS total_customers,
            COUNT(orders.order_id) AS total_orders,
                SUM(orders.revenue) AS total_revenue,
                    ROUND(AVG(orders.revenue), 2) AS average_order_value
                    FROM orders
                    INNER JOIN customers
                        ON orders.customer_id = customers.customer_id
                        GROUP BY customers.region
                        ORDER BY total_revenue DESC;
                        """, conn)
```

### Regional Performance Insight

Ibadan generated the highest revenue at ₦580,000 from 7 orders and 2 customers, with the highest average order value of approximately ₦82,857.

Lagos generated ₦340,000 from 6 orders and 2 customers, while Osogbo generated ₦120,000 from 2 orders and 1 customer.

Ibadan is currently the strongest-performing region in terms of revenue, order volume, and average order value. However, the highest-revenue region is not automatically the best growth opportunity.

Management should investigate customer acquisition, repeat purchasing, and revenue trends in each region before deciding where to focus expansion efforts.

###Customer Behavior & Segmentation

This section segments customers based on their order frequency and total revenue to identify high-value customers and potential growth opportunities.

```python
pd.read_sql_query("""
SELECT
    customer_name,
        COUNT(order_id) AS total_orders,
            SUM(revenue) AS total_revenue,
                CASE
                        WHEN SUM(revenue) >= 250000 THEN 'High Value'
                                WHEN SUM(revenue) >= 180000 THEN 'Medium Value'
                                        ELSE 'Low Value'
                                            END AS customer_segment
                                            FROM orders
                                            INNER JOIN customers
                                                ON orders.customer_id = customers.customer_id
                                                GROUP BY customer_name
                                                ORDER BY total_revenue DESC;
                                                """, conn)
```

### Customer Segmentation Insight

Two customers, Ada and Emeka, were classified as High Value, generating ₦580,000 in combined revenue.

Fatima was classified as Medium Value with ₦195,000 in revenue, while Bola and Chinedu were classified as Low Value.

Although High Value customers represent only 40% of the customer base, they generated approximately 55.77% of total revenue in the dataset.

This indicates that high-value customers are particularly important to the business. Management should prioritize retaining these customers while developing strategies to increase the value of Medium and Low Value customers.
